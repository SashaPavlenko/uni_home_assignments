### Task 3.a
For the given coordinates the following transformation matrix is constructed:

$$
\begin{pmatrix}
    1 \cdot \frac{1}{\sqrt{2}} & 0 & 1 \cdot  \frac{1}{\sqrt{2}} &  1 \\
    0 & 2 \cdot 1 & 0 & -1 \\
    1 \cdot (-\frac{1}{\sqrt{2}}) & 0 & 1 \cdot \frac{1}{\sqrt{2}} & 2 \\
    0 & 0 & 0 & 1
\end{pmatrix}
$$

The transformation matrix from word coordinates back to index coordinates is an inversed transformation matrix

Let us calculate matrices in Python:

```python
import numpy as np

transformation_matrix = np.array([
    [1/np.sqrt(2), 0, 1/np.sqrt(2), 1],
    [0, 2, 0, -1],
    [1/np.sqrt(2), 0, -1/np.sqrt(2), 2],
    [0, 0, 0, 1]
])

print(f"Index to world coordinates mapping: \n{transformation_matrix} \n\n World to index coordinates mapping: \n{np.linalg.inv(transformation_matrix)}")
```

### Task 3.b 

Use the transformation matrix to determine the world space position of the voxel with index
coordinates (2; 1; 5), and the displacement vector in world space that corresponds to incrementing
the first index space coordinate by one voxel, the second one by two voxels, and the third one by
minus one voxel. Which of the answers changes if we move the origin p and why? (2P)

```python
voxel = np.array([2,1,5,1])

e = 1/np.sqrt(2)*np.array([1,0,1])
f = np.array([0,1,0])
g = 1/np.sqrt(2)*np.array([1,0,-1])
ps = [
    np.array([2,1,5,1]),
    np.array([-2,1,5,1]),
    np.array([2,-1,5,1]),
    np.array([2,1,8,1]),
]

for p in ps:
  T = np.array([
    [e[0], f[0], g[0], p[0]],
    [e[1], f[1], g[1], p[1]],
    [e[2], f[2], g[2], p[2]],
    [0, 0, 0, 1]
  ])
  T_inv = np.linalg.inv(T)

  voxel_world_coords = T @ voxel
  dx = np.array([1,0,0,0])
  dy = np.array([0,2,0,0])
  dz = np.array([0,0,-1,0])

  print(T_inv @ (voxel_world_coords + dx))
  print(T_inv @ (voxel_world_coords + dy))
  print(T_inv @ (voxel_world_coords + dz))
  print("\n\n")
```

```python
# Index coordinates of the voxel
voxel = np.array([2, 1, 5, 1])

# Basis vectors and origin
e = 1/np.sqrt(2) * np.array([1, 0, 1])
f = np.array([0, 1, 0])
g = 1/np.sqrt(2) * np.array([1, 0, -1])
p = np.array([1, -1, 2])  # Origin

# Transformation matrix: index → world
T = np.array([
    [e[0], f[0], g[0], p[0]],
    [e[1], f[1], g[1], p[1]],
    [e[2], f[2], g[2], p[2]],
    [0,    0,    0,    1]
])

# World coordinates of voxel (2,1,5)
voxel_world_coords = T @ voxel
print("World coordinates of voxel (2, 1, 5):")
print(np.round(voxel_world_coords[:3], 3))

# Displacement vector for Δ(i, j, k) = (1, 2, -1)
di, dj, dk = 1, 2, -1
displacement = di * e + dj * f + dk * g
print("\nWorld displacement vector for Δ(i=1, j=2, k=-1):")
print(np.round(displacement, 3))

# Explanation
print("\nNote: If origin p changes, world coordinates of the voxel will also change.")
print("However, the displacement vector stays the same since it depends only on e, f, g.")
```

None of the displaced voxel is changed when the origin $p$ is moved, because this vector stands for translation in the world coordinates and is complemented when we multiply with the inversed transformation when going back to the index space.

## Task 4.a

To analyze the accuracy of the **forward finite difference** method, we apply a Taylor expansion of $f(x+h)$ around $ x $:

$$
f(x+h) = f(x) + f'(x)h + \frac{f''(x)}{2}h^2 + \frac{f^{(3)}(x)}{6}h^3 + \mathcal{O}(h^4)
$$

Substitute into the forward difference:

$$
\frac{f(x+h) - f(x)}{h} = f'(x) + \frac{f''(x)}{2}h + \frac{f^{(3)}(x)}{6}h^2 + \mathcal{O}(h^3)
$$

The error starts at order $ h $, so this scheme is **first-order accurate**.

---

## Task 4.b

Now consider the central difference:

$$
\frac{f(x+h) - f(x-h)}{2h}
$$

Using Taylor expansions:

$$
f(x+h) = f(x) + f'(x)h + \frac{f''(x)}{2}h^2 + \frac{f^{(3)}(x)}{6}h^3 + \frac{f^{(4)}(x)}{24}h^4 + \mathcal{O}(h^5)
$$

$$
f(x-h) = f(x) - f'(x)h + \frac{f''(x)}{2}h^2 - \frac{f^{(3)}(x)}{6}h^3 + \frac{f^{(4)}(x)}{24}h^4 + \mathcal{O}(h^5)
$$

Subtract and divide:

$$
\frac{f(x+h) - f(x-h)}{2h} = f'(x) + \frac{f^{(3)}(x)}{6}h^2 + \mathcal{O}(h^4)
$$

This means the central scheme is **second-order accurate**.

## Task 5.a

The standard bilinear interpolation is:

$$
f(x, y) = f_{i,j}(1 - x)(1 - y) + f_{i+1,j}x(1 - y) + f_{i,j+1}(1 - x)y + f_{i+1,j+1}xy
$$

This can be rewritten as:

$$
f(x, y) = \eta \cdot x \cdot y + (f_{i+1,j} - f_{i,j})x + (f_{i,j+1} - f_{i,j})y + f_{i,j}
$$

where:

$$
\eta = f_{i,j} - f_{i+1,j} - f_{i,j+1} + f_{i+1,j+1}
$$

From this, we get:

$$
y_0 = -\frac{f_{i+1,j} - f_{i,j}}{\eta}, \quad x_0 = -\frac{f_{i,j+1} - f_{i,j}}{\eta}
$$

And:

$$
\gamma = f_{i,j} - \eta x_0 y_0 = \frac{f_{i,j} f_{i+1,j+1} - f_{i+1,j} f_{i,j+1}}{\eta}
$$

---

## Task 5.b

We only divide by $\eta$, so division by zero is possible **only if** $\eta = 0$.  
But in the cases where the asymptotic decider is used, $\eta \ne 0$ because:

$$
f_{i,j} + f_{i+1,j+1} \ne f_{i+1,j} + f_{i,j+1}
$$

Hence, division by zero **does not occur** in relevant cases.
